```python


# Storing student names in a list (10 students, one name repeated 3 times)
students_list = ["Amit", "Rohit", "Sneha", "Neha", "Priya","Rahul", "Kiran", "Pooja", "Amit", "Amit"]

unique_students = set(students_list)

print("Original List:")
print(students_list)

print("\nUnique Participants (Set):")
print(unique_students)
```

```python
# Subjects chosen by five students
student1 = {"Maths", "Physics", "Python", "Biology", "English"}
student2 = {"Maths", "Physics", "Python", "English", "Computer"}
student3 = {"Maths", "hindi", "Python", "English", "Economics"}

# Find common subjects
common_subjects = student1 & student2 & student3

print("Common subjects chosen by all students:", common_subjects)
```

```python
# Students in each club
cricket = {"Amit", "Rahul", "Sneha"}
football = {"Rahul", "Sneha", "Neha"}

# Students in both clubs
both_clubs = cricket & football

# Students in only one club
only_one_club = cricket ^ football

print("Students in both clubs:", both_clubs)
print("Students in only one club:", only_one_club)
```

```python

all_students = {"Rahul", "Sneha", "Amit", "Neha", "Priya"}

present_students = {"Rahul", "Sneha", "Priya"}

absent_students = all_students - present_students

print("Absent students:", absent_students)
```

```python

 # Set of course codes
course_codes = {"CS101", "EE102", "ME103", "CE104"}

# Discontinued course
discontinued_course = "ME103"

# Removing the discontinued course using discard()
course_codes.discard(discontinued_course)

print("Updated Course Codes:", course_codes)
```
